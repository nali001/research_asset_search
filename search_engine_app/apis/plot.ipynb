```python
import pandas as pd
import matplotlib.pyplot as plt

# List of filenames to be read

filenames = [
    'queries_response_times_0.csv', 
    'queries_response_times_1.csv',
    'queries_response_times_2.csv',
    'queries_response_times_3.csv',
    'queries_response_times_4.csv',
]

# Dictionary to store the response times from each file
data = {}
labels = [f'Round {i}' for i in range(1,6)]

# Read each CSV file and extract the 'Response Time (seconds)' column
for file in filenames:
    df = pd.read_csv(file)
    data[file] = df['Response Time (seconds)']

# Plotting the boxplot for each file
plt.figure(figsize=(10, 6))
plt.boxplot(data.values(), labels=labels, patch_artist=True)
# plt.xlabel('Files')
plt.ylabel('Response time (seconds)')
# plt.title('Boxplot of Response Times from Multiple Files')
plt.xticks()
plt.tight_layout()
# Save the plot to a PNG file
plot_file_path = 'response_times.png'
plt.savefig(plot_file_path)
plt.show()
```
